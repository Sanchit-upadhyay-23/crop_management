```python
import string
import random
def random_username(length):
    # Generate a random username of specified length
    letters = string.ascii_lowercase  # All lowercase letters
    return ''.join(random.choice(letters) for i in range(length))
# Generate a username of length 8
username = random_username(8)
print(username) 
```
